```python
# Download one/first 5k pool from MGen study
!mkdir -p compress
!cd compress && curl -s https://zenodo.org/records/16920556/files/pools_5k.tar?download=1 \
| tar -xvf - --occurrence pools_5k_0040f80.zip
```

```python
# Decompress .zip archive with the results
!unzip compress/pools_5k_0040f80.zip -d decompress/
```

```python
# Compress results with af3io defaults
!mkdir -p defaults-compress/
!af3io compress decompress/pools_5k_0040f80 defaults-compress/pools_5k_0040f80.af3io.zip
```

```python
# Decompress af3io defaults
!af3io decompress defaults-compress/pools_5k_0040f80.af3io.zip defaults-decompress/
```

```python
# Compare af3io-decompressed to original pool
!diff -r -q decompress/pools_5k_0040f80 defaults-decompress/pools_5k_0040f80
```

```python
# Show disk space used by compressed/uncompressed
!du -hs decompress/
!du -hs defaults-decompress/
!du -hs compress/
!du -hs defaults-compress/
```
