```python
#!wget https://raw.githubusercontent.com/DunbrackLab/IPSAE/refs/heads/main/Example/fold_aurka_0_tpx2_0_full_data_0.json
import gzip, functools, json
import af3io
import numpy as np

from af3io.scoring import chain_pair_reduce, ipsae, ptm_symm
```

```python
path = 'fold_aurka_0_tpx2_0_full_data_0.json'
with open(path, 'rt', encoding="utf-8") as f:
    data = json.load(f)

data.keys()
```

```python
token_chain_ids = np.array(data['token_chain_ids'])
pae = np.array(data['pae'])
ipsae = chain_pair_reduce(functools.partial(ipsae, pae_cutoff=10), token_chain_ids, pae)
ipsae
```
